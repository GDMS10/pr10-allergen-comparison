```python
from scipy.stats import pearsonr
from scipy.stats import spearmanr
import pandas as pd
```

```python
data = pd.read_csv("allergen_comparison.csv")
print(len(data))
```

```python
print(pearsonr(data["seq_identity_pct"], data["rmsd_angstrom"]))
print(spearmanr(data["seq_identity_pct"], data["rmsd_angstrom"]))
```

```python
no_cc = data[~((data["protein_a"] == "Api g 1") & (data["protein_b"] == "Dau c 1"))]
print(len(no_cc))
```

```python
print(pearsonr(no_cc["seq_identity_pct"], no_cc["rmsd_angstrom"]))
print(spearmanr(no_cc["seq_identity_pct"], no_cc["rmsd_angstrom"]))
```

If no real relationship existed, a correlation of this strength would show up 13.9% of the time, and since the cutoff is 5%, this value is insignificant and can't prove the lack of correlation.
